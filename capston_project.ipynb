```python
# ================================================================
# CREDIT RISK ASSESSMENT
# Predicting Credit Card Default Using Machine Learning
# ================================================================
#
# Project workflow:
# 1. Import libraries
# 2. Load dataset
# 3. Understand the dataset
# 4. Data quality checks
# 5. Data cleaning
# 6. Exploratory Data Analysis (EDA)
# 7. Statistical analysis
# 8. Feature engineering
# 9. Train-test split
# 10. Data preprocessing
# 11. Machine learning model development
# 12. Model evaluation
# 13. Confusion matrix
# 14. ROC curves
# 15. Feature importance
# 16. Final conclusions
#
# Target variable:
# default.payment.next.month
#
# Target:
# 0 = No default
# 1 = Default
# ================================================================


# ================================================================
# 1. IMPORT REQUIRED LIBRARIES
# ================================================================

# Numerical calculations
import numpy as np
import pandas as pd

# Data visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Statistical analysis
from scipy.stats import pointbiserialr
from scipy.stats import chi2_contingency

# Machine learning - splitting and validation
from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_val_score
)

# Machine learning - preprocessing
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)

from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer

# Machine learning models
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

# XGBoost
import xgboost as xgb

# Evaluation metrics
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report,
    roc_curve
)

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

# Plot style
sns.set_theme(style="whitegrid")


# ================================================================
# 2. LOAD THE DATASET
# ================================================================

# Change this path if your CSV file is stored somewhere else.
FILE_PATH = "P_5_UCI_Credit_Card.csv"

# Read the CSV file
df = pd.read_csv(FILE_PATH)

# Display the first five rows
print("First five rows:")
display(df.head())


# ================================================================
# 3. BASIC DATASET UNDERSTANDING
# ================================================================

print("\nDataset shape:")
print(df.shape)

print("\nNumber of rows:", df.shape[0])
print("Number of columns:", df.shape[1])


# Display column names
print("\nColumn names:")
print(df.columns.tolist())


# Display data types
print("\nData types:")
print(df.dtypes)


# Display statistical summary
print("\nStatistical summary:")
display(df.describe().T)


# ================================================================
# 4. CHECK DATA QUALITY
# ================================================================

# ------------------------------------------------
# 4.1 Check missing values
# ------------------------------------------------

missing_values = df.isnull().sum()

print("\nMissing values:")
display(
    missing_values[missing_values > 0]
)

print(
    "\nTotal missing values:",
    df.isnull().sum().sum()
)


# ------------------------------------------------
# 4.2 Check duplicate records
# ------------------------------------------------

duplicate_count = df.duplicated().sum()

print("\nNumber of duplicate rows:", duplicate_count)


# ------------------------------------------------
# 4.3 Check unique values
# ------------------------------------------------

print("\nNumber of unique values per column:")

unique_values = df.nunique()

display(unique_values)


# ------------------------------------------------
# 4.4 Check target variable
# ------------------------------------------------

TARGET = "default.payment.next.month"

print("\nTarget variable:", TARGET)

print("\nTarget distribution:")
display(df[TARGET].value_counts())

print("\nTarget distribution (%):")
display(
    df[TARGET].value_counts(normalize=True) * 100
)


# ================================================================
# 5. DATA CLEANING
# ================================================================

# Create a copy so that the original dataset remains unchanged.
data = df.copy()


# ------------------------------------------------
# 5.1 Handle unusual education codes
# ------------------------------------------------

# According to the dataset coding:
#
# 1 = Graduate school
# 2 = University
# 3 = High school
# 4 = Others
#
# Rare values 0, 5 and 6 are grouped with "Other".
#
# We create a NEW column instead of overwriting the
# original EDUCATION column.

data["EDUCATION_GROUP"] = data["EDUCATION"].replace({
    0: 4,
    5: 4,
    6: 4
})


# ------------------------------------------------
# 5.2 Handle unusual marriage code
# ------------------------------------------------

# Marriage:
# 1 = Married
# 2 = Single
# 3 = Others
#
# Rare value 0 is grouped into category 3.

data["MARRIAGE_GROUP"] = data["MARRIAGE"].replace({
    0: 3
})


# ------------------------------------------------
# 5.3 Verify cleaning
# ------------------------------------------------

print("\nEducation values after grouping:")
print(data["EDUCATION_GROUP"].value_counts().sort_index())

print("\nMarriage values after grouping:")
print(data["MARRIAGE_GROUP"].value_counts().sort_index())


# ================================================================
# 6. OUTLIER ANALYSIS
# ================================================================

# Important:
# We do NOT automatically delete outliers.
#
# Large credit limits, bills and payments can represent genuine
# customers rather than errors.
#
# Therefore, the project retains these observations.


# List of important numerical financial variables
financial_columns = [
    "LIMIT_BAL",
    "BILL_AMT1",
    "BILL_AMT2",
    "BILL_AMT3",
    "BILL_AMT4",
    "BILL_AMT5",
    "BILL_AMT6",
    "PAY_AMT1",
    "PAY_AMT2",
    "PAY_AMT3",
    "PAY_AMT4",
    "PAY_AMT5",
    "PAY_AMT6"
]


# Calculate IQR-based outlier counts
outlier_results = []

for column in financial_columns:

    # First quartile
    Q1 = data[column].quantile(0.25)

    # Third quartile
    Q3 = data[column].quantile(0.75)

    # Interquartile range
    IQR = Q3 - Q1

    # Lower and upper limits
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    # Count observations outside the limits
    outliers = (
        (data[column] < lower_limit) |
        (data[column] > upper_limit)
    ).sum()

    outlier_results.append({
        "Variable": column,
        "Outliers": outliers,
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit
    })


outlier_df = pd.DataFrame(outlier_results)

print("\nIQR outlier analysis:")
display(outlier_df)


# ================================================================
# 7. EXPLORATORY DATA ANALYSIS (EDA)
# ================================================================


# ================================================================
# 7.1 TARGET VARIABLE DISTRIBUTION
# ================================================================

plt.figure(figsize=(8, 5))

ax = sns.countplot(
    data=data,
    x=TARGET
)

plt.title(
    "Distribution of Credit Card Default Outcome",
    fontsize=14
)

plt.xlabel("Default Next Month")
plt.ylabel("Number of Customers")

# Add numbers above bars
for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()
plt.show()


# Calculate default rate
default_rate = data[TARGET].mean() * 100

print(
    f"\nOverall default rate: {default_rate:.2f}%"
)


# ================================================================
# 7.2 CREDIT LIMIT DISTRIBUTION
# ================================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    data=data,
    x="LIMIT_BAL",
    bins=40,
    kde=True
)

plt.title("Distribution of Credit Limits")
plt.xlabel("Credit Limit")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()


# ================================================================
# 7.3 CREDIT LIMIT BY DEFAULT STATUS
# ================================================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=data,
    x=TARGET,
    y="LIMIT_BAL",
    showfliers=False
)

plt.title(
    "Credit Limit Distribution by Default Status"
)

plt.xlabel("Default Next Month")
plt.ylabel("Credit Limit")

plt.tight_layout()
plt.show()


# ================================================================
# 7.4 AGE DISTRIBUTION
# ================================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    data=data,
    x="AGE",
    bins=30,
    kde=True
)

plt.title("Age Distribution of Credit Card Clients")
plt.xlabel("Age")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()


# ================================================================
# 7.5 DEFAULT RATE BY SEX
# ================================================================

sex_default = (
    data.groupby("SEX")[TARGET]
    .mean() * 100
)

print("\nDefault rate by sex:")
display(sex_default)


plt.figure(figsize=(7, 5))

sns.barplot(
    x=sex_default.index,
    y=sex_default.values
)

plt.title("Default Rate by Sex")
plt.xlabel("Sex Code")
plt.ylabel("Default Rate (%)")

plt.tight_layout()
plt.show()


# ================================================================
# 7.6 DEFAULT RATE BY EDUCATION
# ================================================================

education_default = (
    data.groupby("EDUCATION_GROUP")[TARGET]
    .mean() * 100
)

print("\nDefault rate by education group:")
display(education_default)


plt.figure(figsize=(8, 5))

sns.barplot(
    x=education_default.index,
    y=education_default.values
)

plt.title("Default Rate by Education Group")
plt.xlabel("Education Group")
plt.ylabel("Default Rate (%)")

plt.tight_layout()
plt.show()


# ================================================================
# 7.7 DEFAULT RATE BY MARRIAGE
# ================================================================

marriage_default = (
    data.groupby("MARRIAGE_GROUP")[TARGET]
    .mean() * 100
)

print("\nDefault rate by marriage group:")
display(marriage_default)


plt.figure(figsize=(8, 5))

sns.barplot(
    x=marriage_default.index,
    y=marriage_default.values
)

plt.title("Default Rate by Marriage Group")
plt.xlabel("Marriage Group")
plt.ylabel("Default Rate (%)")

plt.tight_layout()
plt.show()


# ================================================================
# 8. FEATURE ENGINEERING
# ================================================================

# Payment-status columns
PAY_STATUS_COLUMNS = [
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6"
]


# Statement/bill columns
BILL_COLUMNS = [
    "BILL_AMT1",
    "BILL_AMT2",
    "BILL_AMT3",
    "BILL_AMT4",
    "BILL_AMT5",
    "BILL_AMT6"
]


# Payment amount columns
PAYMENT_COLUMNS = [
    "PAY_AMT1",
    "PAY_AMT2",
    "PAY_AMT3",
    "PAY_AMT4",
    "PAY_AMT5",
    "PAY_AMT6"
]


# ------------------------------------------------
# 8.1 Maximum payment delay
# ------------------------------------------------

# Finds the worst repayment status over the six months.

data["MAX_PAY_DELAY"] = (
    data[PAY_STATUS_COLUMNS]
    .max(axis=1)
)


# ------------------------------------------------
# 8.2 Average payment status
# ------------------------------------------------

# Measures the average repayment condition.

data["AVG_PAY_STATUS"] = (
    data[PAY_STATUS_COLUMNS]
    .mean(axis=1)
)


# ------------------------------------------------
# 8.3 Number of delayed months
# ------------------------------------------------

# Count how many months had a positive repayment-status value.

data["NUM_DELAYED_MONTHS"] = (
    data[PAY_STATUS_COLUMNS] > 0
).sum(axis=1)


# ------------------------------------------------
# 8.4 Total bill over six months
# ------------------------------------------------

data["TOTAL_BILL_6M"] = (
    data[BILL_COLUMNS]
    .sum(axis=1)
)


# ------------------------------------------------
# 8.5 Total payment over six months
# ------------------------------------------------

data["TOTAL_PAYMENT_6M"] = (
    data[PAYMENT_COLUMNS]
    .sum(axis=1)
)


# ------------------------------------------------
# 8.6 Average bill
# ------------------------------------------------

data["AVG_BILL_6M"] = (
    data[BILL_COLUMNS]
    .mean(axis=1)
)


# ------------------------------------------------
# 8.7 Average payment
# ------------------------------------------------

data["AVG_PAYMENT_6M"] = (
    data[PAYMENT_COLUMNS]
    .mean(axis=1)
)


# ------------------------------------------------
# 8.8 Payment-to-bill ratio
# ------------------------------------------------

# A higher value generally indicates greater payment activity
# relative to the amount billed.

data["PAYMENT_TO_BILL_RATIO"] = (
    data["TOTAL_PAYMENT_6M"] /
    (data["TOTAL_BILL_6M"].abs() + 1)
)


# ------------------------------------------------
# 8.9 Average credit utilisation
# ------------------------------------------------

# Approximate utilisation = average bill / credit limit.

data["AVG_UTILIZATION"] = (
    data["AVG_BILL_6M"] /
    (data["LIMIT_BAL"] + 1)
)


# ------------------------------------------------
# 8.10 Recent utilisation
# ------------------------------------------------

data["LAST_MONTH_UTILIZATION"] = (
    data["BILL_AMT1"] /
    (data["LIMIT_BAL"] + 1)
)


# Display newly created variables
engineered_features = [
    "MAX_PAY_DELAY",
    "AVG_PAY_STATUS",
    "NUM_DELAYED_MONTHS",
    "TOTAL_BILL_6M",
    "TOTAL_PAYMENT_6M",
    "AVG_BILL_6M",
    "AVG_PAYMENT_6M",
    "PAYMENT_TO_BILL_RATIO",
    "AVG_UTILIZATION",
    "LAST_MONTH_UTILIZATION"
]

print("\nEngineered features:")
display(data[engineered_features].head())


# ================================================================
# 9. DEFAULT RATE BASED ON PAYMENT DELAY
# ================================================================

# Create groups based on maximum historical delay.

data["DELAY_GROUP"] = pd.cut(
    data["MAX_PAY_DELAY"],
    bins=[
        -np.inf,
        0,
        1,
        2,
        3,
        np.inf
    ],
    labels=[
        "≤ 0",
        "1 month",
        "2 months",
        "3 months",
        "4+ months"
    ]
)


# Calculate default rate for each group
delay_default_rate = (
    data.groupby(
        "DELAY_GROUP",
        observed=False
    )[TARGET]
    .mean() * 100
)


print("\nDefault rate by maximum payment delay:")
display(delay_default_rate)


# Visualization
plt.figure(figsize=(9, 5))

ax = sns.barplot(
    x=delay_default_rate.index,
    y=delay_default_rate.values
)

plt.title(
    "Default Rate by Maximum Historical Payment Delay"
)

plt.xlabel(
    "Maximum Historical Payment Delay"
)

plt.ylabel("Default Rate (%)")

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f"
    )

plt.tight_layout()
plt.show()


# ================================================================
# 10. DEFAULT RATE BY NUMBER OF DELAYED MONTHS
# ================================================================

delayed_month_rate = (
    data.groupby("NUM_DELAYED_MONTHS")[TARGET]
    .mean() * 100
)

plt.figure(figsize=(9, 5))

sns.lineplot(
    x=delayed_month_rate.index,
    y=delayed_month_rate.values,
    marker="o"
)

plt.title(
    "Default Rate vs Number of Delayed Payment Months"
)

plt.xlabel("Number of Delayed Months")
plt.ylabel("Default Rate (%)")

plt.tight_layout()
plt.show()


# ================================================================
# 11. STATISTICAL ANALYSIS
# ================================================================


# ================================================================
# 11.1 Descriptive statistics by target
# ================================================================

important_variables = [
    "LIMIT_BAL",
    "AGE",
    "MAX_PAY_DELAY",
    "NUM_DELAYED_MONTHS",
    "AVG_UTILIZATION",
    "TOTAL_BILL_6M",
    "TOTAL_PAYMENT_6M"
]


group_statistics = (
    data.groupby(TARGET)[important_variables]
    .agg(["mean", "median", "std"])
)


print("\nDescriptive statistics by default status:")
display(group_statistics)


# ================================================================
# 11.2 Point-biserial correlation
# ================================================================

# Point-biserial correlation is appropriate when one variable
# is binary and the other is numerical.

numeric_features = [
    "LIMIT_BAL",
    "AGE",
    "BILL_AMT1",
    "BILL_AMT2",
    "BILL_AMT3",
    "BILL_AMT4",
    "BILL_AMT5",
    "BILL_AMT6",
    "PAY_AMT1",
    "PAY_AMT2",
    "PAY_AMT3",
    "PAY_AMT4",
    "PAY_AMT5",
    "PAY_AMT6",
    "MAX_PAY_DELAY",
    "AVG_PAY_STATUS",
    "NUM_DELAYED_MONTHS",
    "TOTAL_BILL_6M",
    "TOTAL_PAYMENT_6M",
    "AVG_BILL_6M",
    "AVG_PAYMENT_6M",
    "PAYMENT_TO_BILL_RATIO",
    "AVG_UTILIZATION",
    "LAST_MONTH_UTILIZATION"
]


correlation_results = []

for feature in numeric_features:

    r, p = pointbiserialr(
        data[TARGET],
        data[feature]
    )

    correlation_results.append({
        "Feature": feature,
        "Correlation": r,
        "p_value": p
    })


correlation_df = pd.DataFrame(
    correlation_results
)

# Sort according to absolute correlation
correlation_df = correlation_df.sort_values(
    by="Correlation",
    key=lambda x: x.abs(),
    ascending=False
)


print("\nPoint-biserial correlations:")
display(correlation_df.head(15))


# ================================================================
# 11.3 Chi-square tests for categorical variables
# ================================================================

categorical_variables = [
    "SEX",
    "EDUCATION_GROUP",
    "MARRIAGE_GROUP"
]


chi_square_results = []

for variable in categorical_variables:

    # Create contingency table
    contingency_table = pd.crosstab(
        data[variable],
        data[TARGET]
    )

    # Perform chi-square test
    chi2, p, dof, expected = chi2_contingency(
        contingency_table
    )

    # Total number of observations
    n = contingency_table.values.sum()

    # Calculate Cramer's V
    min_dimension = min(
        contingency_table.shape
    )

    cramers_v = np.sqrt(
        chi2 /
        (n * (min_dimension - 1))
    )

    chi_square_results.append({
        "Variable": variable,
        "Chi-square": chi2,
        "df": dof,
        "p-value": p,
        "Cramer's V": cramers_v
    })


chi_square_df = pd.DataFrame(
    chi_square_results
)


print("\nChi-square test results:")
display(chi_square_df)


# ================================================================
# 12. CORRELATION HEATMAP
# ================================================================

heatmap_variables = [
    "LIMIT_BAL",
    "AGE",
    "MAX_PAY_DELAY",
    "NUM_DELAYED_MONTHS",
    "AVG_UTILIZATION",
    "TOTAL_BILL_6M",
    "TOTAL_PAYMENT_6M",
    "PAY_AMT1",
    "PAY_AMT2",
    "PAY_AMT3",
    "PAY_AMT4",
    "PAY_AMT5",
    "PAY_AMT6",
    TARGET
]


correlation_matrix = (
    data[heatmap_variables]
    .corr()
)


plt.figure(figsize=(12, 9))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title(
    "Correlation Heatmap of Important Variables"
)

plt.tight_layout()
plt.show()


# ================================================================
# 13. PREPARE DATA FOR MACHINE LEARNING
# ================================================================


# ------------------------------------------------
# 13.1 Define categorical variables
# ------------------------------------------------

categorical_features = [
    "SEX",
    "EDUCATION_GROUP",
    "MARRIAGE_GROUP"
] + PAY_STATUS_COLUMNS


# ------------------------------------------------
# 13.2 Define numerical variables
# ------------------------------------------------

numerical_features = [
    "LIMIT_BAL",
    "AGE"
] + BILL_COLUMNS + PAYMENT_COLUMNS + [
    "MAX_PAY_DELAY",
    "AVG_PAY_STATUS",
    "NUM_DELAYED_MONTHS",
    "TOTAL_BILL_6M",
    "TOTAL_PAYMENT_6M",
    "AVG_BILL_6M",
    "AVG_PAYMENT_6M",
    "PAYMENT_TO_BILL_RATIO",
    "AVG_UTILIZATION",
    "LAST_MONTH_UTILIZATION"
]


# ------------------------------------------------
# 13.3 Create X and y
# ------------------------------------------------

X = data[
    categorical_features +
    numerical_features
]

y = data[TARGET]


print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)


# ================================================================
# 14. TRAIN-TEST SPLIT
# ================================================================

# 80% of data will be used for training.
# 20% will be reserved for final testing.
#
# stratify=y ensures that both training and testing sets
# retain approximately the same default/non-default ratio.

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))


# ================================================================
# 15. DATA PREPROCESSING
# ================================================================


# ------------------------------------------------
# Numerical preprocessing
# ------------------------------------------------

# Missing numerical values will be replaced by the median.
# Then numerical features will be standardised.

numerical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "scaler",
        StandardScaler()
    )
])


# ------------------------------------------------
# Categorical preprocessing
# ------------------------------------------------

# Missing categorical values are replaced with the most frequent
# category.
#
# One-hot encoding converts categorical variables into numerical
# columns suitable for machine learning.

categorical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="most_frequent")
    ),
    (
        "onehot",
        OneHotEncoder(
            handle_unknown="ignore"
        )
    )
])


# ------------------------------------------------
# Combine both preprocessing pipelines
# ------------------------------------------------

preprocessor = ColumnTransformer([
    (
        "numerical",
        numerical_pipeline,
        numerical_features
    ),
    (
        "categorical",
        categorical_pipeline,
        categorical_features
    )
])


# ================================================================
# 16. DEFINE MACHINE LEARNING MODELS
# ================================================================


# ------------------------------------------------
# Model 1: Logistic Regression
# ------------------------------------------------

# Logistic Regression provides a simple and interpretable
# baseline classification model.

logistic_model = LogisticRegression(
    max_iter=2000,
    class_weight="balanced",
    C=1.0,
    solver="liblinear"
)


# ------------------------------------------------
# Model 2: Random Forest
# ------------------------------------------------

# Random Forest is a tree-based ensemble model.
# It can model nonlinear relationships and interactions.

random_forest_model = RandomForestClassifier(
    n_estimators=400,
    max_depth=14,
    min_samples_leaf=3,
    class_weight="balanced_subsample",
    random_state=42,
    n_jobs=-1
)


# ------------------------------------------------
# Model 3: XGBoost
# ------------------------------------------------

# XGBoost is a gradient-boosted decision-tree algorithm.
#
# scale_pos_weight gives additional importance to the minority
# default class during training.

class_weight_ratio = (
    (y_train == 0).sum() /
    (y_train == 1).sum()
)


xgboost_model = xgb.XGBClassifier(
    n_estimators=400,
    max_depth=3,
    learning_rate=0.04,
    min_child_weight=5,
    subsample=0.85,
    colsample_bytree=0.85,
    reg_lambda=2,

    objective="binary:logistic",

    eval_metric="logloss",

    scale_pos_weight=class_weight_ratio,

    random_state=42,
    n_jobs=-1
)


# ================================================================
# 17. CREATE MACHINE LEARNING PIPELINES
# ================================================================

# A pipeline ensures that preprocessing and model training
# happen together.
#
# This also helps prevent data leakage.

models = {

    "Logistic Regression": Pipeline([
        (
            "preprocessing",
            preprocessor
        ),
        (
            "model",
            logistic_model
        )
    ]),

    "Random Forest": Pipeline([
        (
            "preprocessing",
            preprocessor
        ),
        (
            "model",
            random_forest_model
        )
    ]),

    "XGBoost": Pipeline([
        (
            "preprocessing",
            preprocessor
        ),
        (
            "model",
            xgboost_model
        )
    ])
}


# ================================================================
# 18. TRAIN AND EVALUATE ALL MODELS
# ================================================================

results = []

trained_models = {}

predicted_probabilities = {}


for model_name, pipeline in models.items():

    print("\n" + "=" * 60)
    print("Training:", model_name)
    print("=" * 60)

    # Train model
    pipeline.fit(
        X_train,
        y_train
    )

    # Save trained model
    trained_models[model_name] = pipeline

    # Generate probability predictions
    probabilities = pipeline.predict_proba(
        X_test
    )[:, 1]

    predicted_probabilities[
        model_name
    ] = probabilities

    # Convert probabilities into binary predictions
    predictions = (
        probabilities >= 0.50
    ).astype(int)

    # Calculate evaluation metrics
    accuracy = accuracy_score(
        y_test,
        predictions
    )

    precision = precision_score(
        y_test,
        predictions
    )

    recall = recall_score(
        y_test,
        predictions
    )

    f1 = f1_score(
        y_test,
        predictions
    )

    roc_auc = roc_auc_score(
        y_test,
        probabilities
    )

    pr_auc = average_precision_score(
        y_test,
        probabilities
    )

    # Store results
    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1,
        "ROC-AUC": roc_auc,
        "PR-AUC": pr_auc
    })


# Convert results to DataFrame
results_df = pd.DataFrame(results)


# Display model comparison
print("\nMODEL COMPARISON")
display(
    results_df.round(4)
)


# ================================================================
# 19. VISUALIZE MODEL COMPARISON
# ================================================================

metrics_to_plot = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1-Score",
    "ROC-AUC",
    "PR-AUC"
]


for metric in metrics_to_plot:

    plt.figure(figsize=(8, 5))

    sns.barplot(
        data=results_df,
        x="Model",
        y=metric
    )

    plt.title(
        f"Model Comparison - {metric}"
    )

    plt.ylabel(metric)
    plt.xlabel("Machine Learning Model")

    plt.xticks(rotation=15)

    plt.tight_layout()
    plt.show()


# ================================================================
# 20. SELECT FINAL MODEL
# ================================================================

# For this project, XGBoost is used as the final model because
# it provides strong ranking performance and nonlinear modelling
# capability.

FINAL_MODEL_NAME = "XGBoost"

final_model = trained_models[
    FINAL_MODEL_NAME
]

final_probabilities = predicted_probabilities[
    FINAL_MODEL_NAME
]


# Convert probabilities into predictions
final_predictions = (
    final_probabilities >= 0.50
).astype(int)


# Retrieve final metrics
final_results = results_df[
    results_df["Model"] == FINAL_MODEL_NAME
].iloc[0]


print("\nFINAL MODEL:", FINAL_MODEL_NAME)

print(
    f"Accuracy : {final_results['Accuracy']:.4f}"
)

print(
    f"Precision: {final_results['Precision']:.4f}"
)

print(
    f"Recall   : {final_results['Recall']:.4f}"
)

print(
    f"F1-Score : {final_results['F1-Score']:.4f}"
)

print(
    f"ROC-AUC  : {final_results['ROC-AUC']:.4f}"
)

print(
    f"PR-AUC   : {final_results['PR-AUC']:.4f}"
)


# ================================================================
# 21. CLASSIFICATION REPORT
# ================================================================

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        final_predictions,
        target_names=[
            "No Default",
            "Default"
        ]
    )
)


# ================================================================
# 22. CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(
    y_test,
    final_predictions
)


print("\nConfusion Matrix:")
print(cm)


plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt=",d",
    cmap="Blues",
    xticklabels=[
        "Predicted No Default",
        "Predicted Default"
    ],
    yticklabels=[
        "Actual No Default",
        "Actual Default"
    ]
)

plt.title(
    "XGBoost Confusion Matrix"
)

plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.tight_layout()
plt.show()


# ================================================================
# 23. ROC CURVE COMPARISON
# ================================================================

plt.figure(figsize=(8, 6))


for model_name, probabilities in predicted_probabilities.items():

    # Calculate False Positive Rate and True Positive Rate
    fpr, tpr, thresholds = roc_curve(
        y_test,
        probabilities
    )

    # Calculate AUC
    auc_score = roc_auc_score(
        y_test,
        probabilities
    )

    # Plot ROC curve
    plt.plot(
        fpr,
        tpr,
        label=f"{model_name} (AUC = {auc_score:.3f})"
    )


# Random classifier baseline
plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)


plt.title(
    "ROC Curves for Machine Learning Models"
)

plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.legend()

plt.tight_layout()
plt.show()


# ================================================================
# 24. FEATURE IMPORTANCE - XGBOOST
# ================================================================

# Extract the trained XGBoost model
xgb_pipeline = trained_models[
    "XGBoost"
]


# Extract the preprocessing component
fitted_preprocessor = (
    xgb_pipeline
    .named_steps["preprocessing"]
)


# Extract the trained XGBoost model
fitted_xgb = (
    xgb_pipeline
    .named_steps["model"]
)


# Get transformed feature names
feature_names = (
    fitted_preprocessor
    .get_feature_names_out()
)


# Get XGBoost feature importance
importance_values = (
    fitted_xgb
    .feature_importances_
)


# Create DataFrame
feature_importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importance_values
})


# Sort by importance
feature_importance_df = (
    feature_importance_df
    .sort_values(
        by="Importance",
        ascending=False
    )
)


print("\nTop 20 model features:")
display(
    feature_importance_df.head(20)
)


# ================================================================
# 25. PLOT TOP 20 FEATURES
# ================================================================

top_features = (
    feature_importance_df
    .head(20)
    .sort_values(
        by="Importance"
    )
)


plt.figure(figsize=(10, 8))

plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)

plt.title(
    "Top 20 XGBoost Feature Importances"
)

plt.xlabel(
    "Feature Importance"
)

plt.ylabel(
    "Feature"
)

plt.tight_layout()
plt.show()


# ================================================================
# 26. AGGREGATE FEATURE IMPORTANCE
# ================================================================

# One-hot encoding creates several columns for categorical
# variables.
#
# Here we aggregate their importance back to the original
# variable names.

aggregated_importance = {}


for feature, importance in zip(
    feature_names,
    importance_values
):

    # Remove preprocessing prefix
    original_feature = feature.split(
        "__",
        1
    )[-1]

    matched_feature = None

    # Check categorical variables
    for categorical_feature in categorical_features:

        if original_feature.startswith(
            categorical_feature + "_"
        ):

            matched_feature = (
                categorical_feature
            )

            break


    # If not categorical, retain original name
    if matched_feature is None:

        matched_feature = (
            original_feature
        )


    # Add importance
    aggregated_importance[
        matched_feature
    ] = (
        aggregated_importance.get(
            matched_feature,
            0
        )
        + importance
    )


# Convert to Series
aggregated_importance = pd.Series(
    aggregated_importance
).sort_values(
    ascending=False
)


print("\nAggregated feature importance:")
display(
    aggregated_importance.head(20)
)


# ================================================================
# 27. PLOT AGGREGATED FEATURE IMPORTANCE
# ================================================================

top_aggregated = (
    aggregated_importance
    .head(15)
    .sort_values()
)


plt.figure(figsize=(9, 7))

plt.barh(
    top_aggregated.index,
    top_aggregated.values
)

plt.title(
    "Top Features Used by XGBoost"
)

plt.xlabel(
    "Aggregated Feature Importance"
)

plt.ylabel(
    "Feature"
)

plt.tight_layout()
plt.show()


# ================================================================
# 28. TRAINING VS TESTING PERFORMANCE
# ================================================================

# This section checks for potential overfitting.

train_probability = final_model.predict_proba(
    X_train
)[:, 1]


train_prediction = (
    train_probability >= 0.50
).astype(int)


train_roc_auc = roc_auc_score(
    y_train,
    train_probability
)

test_roc_auc = roc_auc_score(
    y_test,
    final_probabilities
)


train_pr_auc = average_precision_score(
    y_train,
    train_probability
)

test_pr_auc = average_precision_score(
    y_test,
    final_probabilities
)


print("\nTraining vs Testing Performance")

print(
    f"Training ROC-AUC: {train_roc_auc:.4f}"
)

print(
    f"Testing ROC-AUC : {test_roc_auc:.4f}"
)

print(
    f"Training PR-AUC : {train_pr_auc:.4f}"
)

print(
    f"Testing PR-AUC  : {test_pr_auc:.4f}"
)


# ================================================================
# 29. 5-FOLD CROSS VALIDATION
# ================================================================

# Cross-validation provides an additional estimate of how stable
# the model's performance is.

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


cv_scores = cross_val_score(
    final_model,
    X_train,
    y_train,
    cv=cv,
    scoring="roc_auc"
)


print("\n5-Fold Cross-Validation ROC-AUC Scores:")

for i, score in enumerate(
    cv_scores,
    start=1
):

    print(
        f"Fold {i}: {score:.4f}"
    )


print(
    "\nMean CV ROC-AUC:",
    round(cv_scores.mean(), 4)
)

print(
    "Standard deviation:",
    round(cv_scores.std(), 4)
)


# ================================================================
# 30. BORROWER-LEVEL RISK PREDICTION EXAMPLE
# ================================================================

# The model produces a probability between 0 and 1.
#
# For example:
#
# 0.10 = relatively low predicted risk
# 0.50 = threshold used in this project
# 0.80 = high predicted probability
#
# IMPORTANT:
# These are model probabilities, not guarantees that a borrower
# will or will not default.


# Create a result DataFrame for the test set

prediction_results = X_test.copy()

prediction_results["Actual_Default"] = (
    y_test.values
)

prediction_results["Predicted_Default_Probability"] = (
    final_probabilities
)

prediction_results["Predicted_Default"] = (
    final_predictions
)


print("\nSample borrower predictions:")
display(
    prediction_results[
        [
            "Actual_Default",
            "Predicted_Default_Probability",
            "Predicted_Default"
        ]
    ].head(20)
)


# ================================================================
# 31. RISK GROUP VISUALIZATION
# ================================================================

# Divide predicted probabilities into three broad analytical
# groups.
#
# These groups are used only for visualization and interpretation.
# They are NOT official credit-risk categories.

prediction_results["Risk_Group"] = pd.cut(
    prediction_results[
        "Predicted_Default_Probability"
    ],
    bins=[
        -np.inf,
        0.30,
        0.60,
        np.inf
    ],
    labels=[
        "Lower predicted probability",
        "Medium predicted probability",
        "Higher predicted probability"
    ]
)


risk_distribution = (
    prediction_results["Risk_Group"]
    .value_counts()
    .sort_index()
)


print("\nPredicted probability groups:")
display(risk_distribution)


plt.figure(figsize=(9, 5))

sns.barplot(
    x=risk_distribution.index,
    y=risk_distribution.values
)

plt.title(
    "Distribution of Predicted Default Probability Groups"
)

plt.xlabel(
    "Predicted Probability Group"
)

plt.ylabel(
    "Number of Customers"
)

plt.xticks(rotation=15)

plt.tight_layout()
plt.show()


# ================================================================
# 32. EXPORT RESULTS
# ================================================================

# Save model comparison
results_df.to_csv(
    "model_comparison.csv",
    index=False
)


# Save test-set predictions
prediction_results.to_csv(
    "test_set_predictions.csv",
    index=False
)


# Save cleaned / engineered dataset
data.to_csv(
    "credit_card_default_cleaned_features.csv",
    index=False
)


# Save feature importance
feature_importance_df.to_csv(
    "xgboost_feature_importance.csv",
    index=False
)


print("\nAll analysis results have been exported.")


# ================================================================
# 33. FINAL PROJECT SUMMARY
# ================================================================

print("\n")
print("=" * 70)
print("FINAL PROJECT SUMMARY")
print("=" * 70)

print(
    f"Dataset size       : {data.shape[0]:,} rows × {df.shape[1]} columns"
)

print(
    f"Default rate       : {default_rate:.2f}%"
)

print(
    f"Final model        : {FINAL_MODEL_NAME}"
)

print(
    f"Accuracy           : {final_results['Accuracy']:.4f}"
)

print(
    f"Precision          : {final_results['Precision']:.4f}"
)

print(
    f"Recall             : {final_results['Recall']:.4f}"
)

print(
    f"F1-score           : {final_results['F1-Score']:.4f}"
)

print(
    f"ROC-AUC            : {final_results['ROC-AUC']:.4f}"
)

print(
    f"PR-AUC             : {final_results['PR-AUC']:.4f}"
)

print(
    f"CV ROC-AUC         : {cv_scores.mean():.4f}"
    f" ± {cv_scores.std():.4f}"
)

print("=" * 70)


# ================================================================
# 34. KEY INTERPRETATION
# ================================================================

print("""
KEY INTERPRETATION
-------------------

1. Historical repayment behaviour is strongly associated with
   next-month default.

2. Customers with more severe or more frequent repayment delays
   tend to have higher observed default rates.

3. Credit limit and payment activity also contain predictive
   information, although their relationships are weaker than the
   repayment-status variables.

4. The XGBoost model captures nonlinear relationships and
   interactions between borrower characteristics and repayment
   behaviour.

5. The model is not perfect. False positives and false negatives
   remain, so it should be treated as decision-support rather than
   an automatic lending decision system.

6. The model was evaluated on unseen test data and additionally
   checked using five-fold cross-validation.

7. Future work should include temporal validation, additional
   financial variables, probability calibration and fairness/
   governance analysis before any real-world deployment.
""")
```
